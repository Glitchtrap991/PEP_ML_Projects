```python
import time

start_time = time.time()

#Trial using the Medium YOLO v26 Model

model = YOLO("yolo26m.pt")

results = model.train(
     data="dataset.yaml",
     epochs=50,
     save=True,
     save_period=5,
     project="IEDyolo_960px_medium",
     imgsz=960,
     batch=-1,
     device=0,
     hsv_h=0.015,
     hsv_s=0.4,
     hsv_v=0.3,

     degrees=5.0,
     translate=0.1,
     scale=0.3,
     shear=2.0,

     fliplr=0.5,
     flipud=0.0,

     mosaic=0.5,
     mixup=0.0,
    
     workers=16,
     amp=True
 )

end_time = time.time()

print(f"Time taken for the training is : {start_time-end_time}")
```

```python
print("\n=============TRAINING METRICS==============\n")
print("Precision:  ", results.box.mp)
print("Recall:     ", results.box.mr)
print("mAP@50:     ", results.box.map50)
print("mAP@75:     ", results.box.map75)
print("mAP@50-95:  ", results.box.map)
```
